Importing the necessary libs

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#loading the dataset
df = pd.read_csv('/Users/wins/ML_Projects/LSTM_CNN_Desalination/Data/UAE_Pump_Data.csv')
```

```python
#basic checks
print("Shape of dataset: ", df.shape)
print("Column names: ", df.columns)
print("Data types:\n", df.dtypes)
print("Number of duplicates:", df.duplicated().sum())
print("Summary statistics:\n", df.describe())
```

```python
#checking the distribution of numerical features
import seaborn as sns
import matplotlib.pyplot as plt

# Plot histograms for numerical features
numerical_features = ['Vibration (mm/s²)', 'Vibration Frequency (Hz)', 'Rotational Speed (RPM)', 
                      'Flow Rate (L/min)', 'Inlet Pressure (bar)', 'Outlet Pressure (bar)', 
                      'Motor Temperature (°C)', 'Pump Bearing Temperature (°C)', 'Current (Amps)', 
                      'Voltage (Volts)', 'Power Consumption (kW)', 'Water Conductivity (µS/cm)', 
                      'Salinity (%)', 'pH Level', 'Ambient Temperature (°C)', 
                      'Time Since Last Maintenance (days)']

for feature in numerical_features:
    plt.figure(figsize=(6, 4))
    sns.histplot(df[feature], kde=True)
    plt.title(f'Distribution of {feature}')
    plt.show()
```

```python
#checking the distribution of failure

# Count the occurrences of each unique value in the 'Failure' column
failure_counts = df['Failure'].value_counts()

# Display the result
print("Number of failures:\n", failure_counts)

import seaborn as sns
import matplotlib.pyplot as plt

# Create a bar plot
plt.figure(figsize=(6, 4))
sns.countplot(x='Failure', data=df)
plt.title('Distribution of Failures')
plt.xlabel('Failure Type')
plt.ylabel('Count')
plt.show()
```
